```python
# =========================================================
# Fig. 5(a): Conventional baselines vs PB-MMAE
# Final refined version — CAE label on right
# =========================================================

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Paths and authoritative locked results
# ---------------------------------------------------------
PROJECT = Path(r"C:\project 3\Paper1_SSL")
LOCK_DIR = PROJECT / "03_outputs" / "final_results_lock"

baseline_df = pd.read_csv(
    LOCK_DIR / "Paper1_FINAL_lock_baseline_models.csv"
)

primary_df = pd.read_csv(
    LOCK_DIR / "Paper1_FINAL_lock_primary_PBMMAE_13x13.csv"
)

primary_metrics = dict(
    zip(primary_df["Metric"], primary_df["Value"])
)

# ---------------------------------------------------------
# Build plotting table
# ---------------------------------------------------------
baseline_plot = baseline_df[
    ["Method", "Silhouette", "DBI"]
].copy()

pbmmae_row = pd.DataFrame(
    [{
        "Method": "PB-MMAE (13×13)",
        "Silhouette": primary_metrics["Silhouette"],
        "DBI": primary_metrics["DBI"]
    }]
)

plot_df = pd.concat(
    [baseline_plot, pbmmae_row],
    ignore_index=True
)

display(plot_df)

# ---------------------------------------------------------
# Plot
# ---------------------------------------------------------
fig, ax = plt.subplots(figsize=(7.8, 5.8))

for _, row in plot_df.iterrows():

    method = row["Method"]
    x = row["DBI"]
    y = row["Silhouette"]

    if method == "PB-MMAE (13×13)":
        ax.scatter(
            x,
            y,
            s=190,
            marker="*",
            zorder=5
        )
    else:
        ax.scatter(
            x,
            y,
            s=95,
            marker="o",
            zorder=4
        )

# ---------------------------------------------------------
# Labels
# ---------------------------------------------------------

# NMF
r = plot_df.loc[
    plot_df["Method"] == "NMF + k-means"
].iloc[0]

ax.annotate(
    "NMF + k-means",
    (r["DBI"], r["Silhouette"]),
    xytext=(8, 3),
    textcoords="offset points",
    fontsize=10,
    fontweight="bold",
    ha="left",
    va="center"
)

# SOM
r = plot_df.loc[
    plot_df["Method"] == "SOM"
].iloc[0]

ax.annotate(
    "SOM",
    (r["DBI"], r["Silhouette"]),
    xytext=(8, 8),
    textcoords="offset points",
    fontsize=10,
    fontweight="bold",
    ha="left",
    va="center"
)

# PCA
r = plot_df.loc[
    plot_df["Method"] == "PCA + k-means"
].iloc[0]

ax.annotate(
    "PCA + k-means",
    (r["DBI"], r["Silhouette"]),
    xytext=(8, -12),
    textcoords="offset points",
    fontsize=10,
    fontweight="bold",
    ha="left",
    va="center"
)

# Conventional CAE — label on RIGHT
r = plot_df.loc[
    plot_df["Method"] == "Conventional CAE + k-means"
].iloc[0]

ax.annotate(
    "Conventional CAE + k-means",
    (r["DBI"], r["Silhouette"]),
    xytext=(10, 7),
    textcoords="offset points",
    fontsize=9.5,
    fontweight="bold",
    ha="left",
    va="bottom"
)

# PB-MMAE
r = plot_df.loc[
    plot_df["Method"] == "PB-MMAE (13×13)"
].iloc[0]

ax.annotate(
    "PB-MMAE (13×13)",
    (r["DBI"], r["Silhouette"]),
    xytext=(-6, 13),
    textcoords="offset points",
    fontsize=10,
    fontweight="bold",
    ha="center",
    va="bottom"
)

# ---------------------------------------------------------
# Minimal interpretation note
# ---------------------------------------------------------
ax.text(
    0.98,
    0.96,
    "Internal separation only",
    transform=ax.transAxes,
    ha="right",
    va="top",
    fontsize=10,
    fontweight="bold"
)

# ---------------------------------------------------------
# Panel label
# ---------------------------------------------------------
ax.text(
    0.015,
    0.985,
    "(a)",
    transform=ax.transAxes,
    fontsize=14,
    fontweight="bold",
    va="top",
    ha="left"
)

# ---------------------------------------------------------
# Axes
# ---------------------------------------------------------
ax.set_xlabel(
    "Davies–Bouldin index (lower is better)",
    fontsize=12,
    fontweight="bold"
)

ax.set_ylabel(
    "Silhouette coefficient (higher is better)",
    fontsize=12,
    fontweight="bold"
)

ax.tick_params(
    axis="both",
    labelsize=11,
    width=1.2,
    length=5
)

for tick in ax.get_xticklabels() + ax.get_yticklabels():
    tick.set_fontweight("bold")

# Extra room on right for CAE label
ax.set_xlim(0.4, 4.05)

ax.set_ylim(
    0.0,
    plot_df["Silhouette"].max() + 0.10
)

ax.grid(
    linestyle="--",
    alpha=0.22
)

for spine in ax.spines.values():
    spine.set_linewidth(1.0)

plt.tight_layout()
plt.show()
```

```python
# =========================================================
# Fig. 5(b): Patch-scale robustness
# 13×13 primary vs 9×9 sensitivity
# Final refined version
# =========================================================

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Paths
# ---------------------------------------------------------
PROJECT = Path(r"C:\project 3\Paper1_SSL")
LOCK_DIR = PROJECT / "03_outputs" / "final_results_lock"

PRIMARY_FILE = LOCK_DIR / "Paper1_FINAL_lock_primary_PBMMAE_13x13.csv"
SENS9_FILE = LOCK_DIR / "Paper1_FINAL_lock_9x9_sensitivity.csv"
HIERARCHY_FILE = LOCK_DIR / "Paper1_FINAL_lock_hierarchical_stability.csv"

# ---------------------------------------------------------
# Load authoritative locked results
# ---------------------------------------------------------
primary_df = pd.read_csv(PRIMARY_FILE)
sens9_df = pd.read_csv(SENS9_FILE)
hierarchy_df = pd.read_csv(HIERARCHY_FILE)

primary_metrics = dict(
    zip(primary_df["Metric"], primary_df["Value"])
)

sens9_metrics = dict(
    zip(sens9_df["Metric"], sens9_df["Value"])
)

# ---------------------------------------------------------
# Extract hierarchy rows
# ---------------------------------------------------------
h13 = hierarchy_df.loc[
    hierarchy_df["Configuration"] == "Primary 13x13"
].iloc[0]

h9 = hierarchy_df.loc[
    hierarchy_df["Configuration"] == "9x9 sensitivity"
].iloc[0]

# ---------------------------------------------------------
# Comparable metrics
# ---------------------------------------------------------
metrics = [
    "Mean ARI",
    "Mean NMI",
    "Within consensus",
    "4-domain mean agreement",
    "4-domain full agreement"
]

values_13 = [
    primary_metrics["Mean ARI"],
    primary_metrics["Mean NMI"],
    primary_metrics["Within-consensus score"],
    h13["Mean_agreement"],
    h13["Full_agreement"]
]

values_9 = [
    sens9_metrics["Mean ARI"],
    sens9_metrics["Mean NMI"],
    sens9_metrics["Within-consensus score"],
    h9["Mean_agreement"],
    h9["Full_agreement"]
]

# ---------------------------------------------------------
# Plot
# ---------------------------------------------------------
x = np.arange(len(metrics))
width = 0.34

fig, ax = plt.subplots(figsize=(8.2, 5.6))

bars13 = ax.bar(
    x - width/2,
    values_13,
    width,
    label="13×13 primary"
)

bars9 = ax.bar(
    x + width/2,
    values_9,
    width,
    label="9×9 sensitivity"
)

# ---------------------------------------------------------
# Value labels
# ---------------------------------------------------------
for bars in [bars13, bars9]:
    for bar in bars:

        h = bar.get_height()

        ax.text(
            bar.get_x() + bar.get_width()/2,
            h + 0.012,
            f"{h:.3f}",
            ha="center",
            va="bottom",
            fontsize=9,
            fontweight="bold"
        )

# ---------------------------------------------------------
# Tie-fraction annotation
# ---------------------------------------------------------
ax.text(
    0.98,
    0.94,
    f"4-domain tie fraction:\n"
    f"13×13 = {h13['Tie_fraction']:.3f}\n"
    f"9×9 = {h9['Tie_fraction']:.3f}",
    transform=ax.transAxes,
    ha="right",
    va="top",
    fontsize=9.5,
    fontweight="bold"
)

# ---------------------------------------------------------
# Panel label
# ---------------------------------------------------------
ax.text(
    0.015,
    0.985,
    "(b)",
    transform=ax.transAxes,
    fontsize=14,
    fontweight="bold",
    va="top",
    ha="left"
)

# ---------------------------------------------------------
# Formatting
# ---------------------------------------------------------
ax.set_ylabel(
    "Score",
    fontsize=12,
    fontweight="bold"
)

ax.set_xticks(x)

ax.set_xticklabels(
    metrics,
    rotation=25,
    ha="right",
    fontsize=10,
    fontweight="bold"
)

ax.tick_params(
    axis="y",
    labelsize=11,
    width=1.2,
    length=5
)

for tick in ax.get_yticklabels():
    tick.set_fontweight("bold")

ax.set_ylim(0, 0.92)

ax.legend(
    frameon=False,
    fontsize=10,
    loc="upper left",
    bbox_to_anchor=(0.08, 0.98)
)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.22
)

for spine in ax.spines.values():
    spine.set_linewidth(1.0)

plt.tight_layout()
plt.show()
```

```python
# =========================================================
# Fig. 5(c): Ablation analysis
# Final refined version
# =========================================================

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Paths
# ---------------------------------------------------------
PROJECT = Path(r"C:\project 3\Paper1_SSL")
LOCK_DIR = PROJECT / "03_outputs" / "final_results_lock"

ABLATION_FILE = LOCK_DIR / "Paper1_FINAL_lock_ablations.csv"

# ---------------------------------------------------------
# Load authoritative locked results
# ---------------------------------------------------------
ablation_df = pd.read_csv(ABLATION_FILE)

# ---------------------------------------------------------
# Short display names
# ---------------------------------------------------------
name_map = {
    "Full PB-MMAE":
        "Full PB-MMAE",

    "Ablation A — no physics-balanced loss":
        "No physics balance",

    "Ablation B — no whole-modality masking":
        "No modality masking",

    "Ablation C — remove MAG_LD":
        "Remove MAG_LD"
}

plot_df = ablation_df.copy()
plot_df["Display"] = plot_df["Experiment"].map(name_map)

display(
    plot_df[
        [
            "Display",
            "ARI",
            "NMI",
            "Within_consensus",
            "PAC"
        ]
    ]
)

# ---------------------------------------------------------
# Higher-is-better metrics
# ---------------------------------------------------------
metrics = [
    "ARI",
    "NMI",
    "Within consensus"
]

metric_cols = [
    "ARI",
    "NMI",
    "Within_consensus"
]

x = np.arange(len(plot_df))
width = 0.22

# Slightly taller figure to accommodate PAC strip
fig, ax = plt.subplots(figsize=(8.4, 6.1))

# ---------------------------------------------------------
# Bars
# ---------------------------------------------------------
for i, (label, col) in enumerate(zip(metrics, metric_cols)):

    bars = ax.bar(
        x + (i - 1) * width,
        plot_df[col],
        width,
        label=label
    )

    for bar in bars:

        h = bar.get_height()

        ax.text(
            bar.get_x() + bar.get_width()/2,
            h + 0.010,
            f"{h:.3f}",
            ha="center",
            va="bottom",
            fontsize=8.7,
            fontweight="bold"
        )

# ---------------------------------------------------------
# Panel label
# ---------------------------------------------------------
ax.text(
    0.015,
    0.985,
    "(c)",
    transform=ax.transAxes,
    fontsize=14,
    fontweight="bold",
    va="top",
    ha="left"
)

# ---------------------------------------------------------
# Formatting
# ---------------------------------------------------------
ax.set_ylabel(
    "Score",
    fontsize=12,
    fontweight="bold"
)

ax.set_xticks(x)

ax.set_xticklabels(
    plot_df["Display"],
    rotation=18,
    ha="right",
    fontsize=10,
    fontweight="bold"
)

ax.tick_params(
    axis="y",
    labelsize=11,
    width=1.2,
    length=5
)

for tick in ax.get_yticklabels():
    tick.set_fontweight("bold")

ax.set_ylim(0, 0.75)

ax.legend(
    frameon=False,
    fontsize=10,
    loc="upper center",
    bbox_to_anchor=(0.50, 0.985),
    ncol=3,
    columnspacing=1.8,
    handlelength=1.8
)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.22
)

for spine in ax.spines.values():
    spine.set_linewidth(1.0)

# ---------------------------------------------------------
# PAC summary strip below axes
# ---------------------------------------------------------
pac_labels = [
    f"Full: {plot_df.loc[0, 'PAC']:.3f}",
    f"No physics balance: {plot_df.loc[1, 'PAC']:.3f}",
    f"No modality masking: {plot_df.loc[2, 'PAC']:.3f}",
    f"Remove MAG_LD: {plot_df.loc[3, 'PAC']:.3f}"
]

pac_text = (
    "PAC (lower is better):   "
    + "   |   ".join(pac_labels)
)

fig.text(
    0.5,
    0.015,
    pac_text,
    ha="center",
    va="bottom",
    fontsize=9.3,
    fontweight="bold"
)

# Leave room at bottom for PAC strip
plt.tight_layout(rect=[0, 0.07, 1, 1])

plt.show()
```
